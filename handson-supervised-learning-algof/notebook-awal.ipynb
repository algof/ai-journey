## 1. Problem Definition

### 1.1 Overview

This dataset simulates user behavior on an e-commerce website using Google Analytics session-level data. Each row represents a unique session, with various behavioral, traffic source, and device-related attributes. The goal is to predict whether a user will make a purchase on a return visit, based on information from their current session.

### 1.2 Objective

Your task is to build a model that predicts the binary outcome:

Will the user make a purchase if they return in a later session?

This can help businesses understand visitor intent, personalize user experience, and retarget high-quality traffic.

## 3. Exploratory Data Analysis

### 3.1 Feature

- **unique_session_id**  
  Unique identifier for each session (based on fullVisitorId and visitId).
  
- **bounces**  
  Indicates if the session ended with only one pageview.
  
- **time_on_site**  
  Total time spent on the site during the session (in seconds).
  
- **pageviews**  
  Number of pages viewed during the session.
  
- **hits**  
  Total number of user interactions during the session.
  
- **session_quality_dim**  
  Proprietary quality score (1–100) estimating purchase intent.
  
- **latest_ecommerce_progress**  
  The furthest stage reached in the e-commerce funnel.
  
- **avg_time_per_page**  
  Average time per page (calculated as time_on_site / pageviews).
  
- **source**  
  The source of the traffic (e.g., google, bing, direct).
  
- **medium**  
  The medium of the traffic (e.g., organic, cpc, referral).
  
- **channelGrouping**  
  High-level channel category (e.g., Referral, Organic Search, Direct).
  
- **deviceCategory**  
  The category of the device used (desktop, tablet, mobile).
  
- **operatingSystem**  
  The category of the device operating system used.
  
- **browser**  
  The category of the device browser used.
  
- **country**  
  Country from where the user accessed the site.
  
- **city**  
  City from where the user accessed the site.

### 4.2 Decision Tree

````python
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score

X = train_df.drop(['will_buy_on_return_visit'], axis=1)  # Fitur
y = train_df['will_buy_on_return_visit']  # Target

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

model = DecisionTreeClassifier(random_state=42)

model.fit(X_train, y_train)

y_pred = model.predict(X_test)

accuracy = accuracy_score(y_test, y_pred)
print(f"Accuracy: {accuracy * 100:.2f}%")
````

### 4.11 LightGBM

````python
import lightgbm as lgb
from sklearn.metrics import accuracy_score

# Membuat dataset LightGBM
train_data = lgb.Dataset(X_train, label=y_train)
test_data = lgb.Dataset(X_test, label=y_test, reference=train_data)

# Menentukan parameter LightGBM
params = {
    'objective': 'binary',  # Klasifikasi biner
    'metric': 'binary_error',  # Menggunakan binary_error sebagai metrik evaluasi
    'num_leaves': 31,  # Jumlah daun dalam setiap pohon
    'learning_rate': 0.07,  # Kecepatan pembelajaran 7
    'n_estimators': 100  # Jumlah estimators (pohon)
}

# Melatih model dengan LightGBM
gbm = lgb.train(params, train_data, valid_sets=[test_data])

# Memprediksi data uji
y_pred = gbm.predict(X_test, num_iteration=gbm.best_iteration)

# Mengubah probabilitas menjadi prediksi biner
y_pred_binary = (y_pred >= 0.5).astype(int)

# Menghitung akurasi
accuracy = accuracy_score(y_test, y_pred_binary)
print(f"Akurasi model: {accuracy * 100:.2f}%")
````

````python
test_pred = gbm.predict(test_df, num_iteration=gbm.best_iteration)

test_pred_binary = (test_pred >= 0.5).astype(int)

df_test_pred = pd.DataFrame(test_pred_binary)

df_test_pred.to_csv(f"submission-{2}.csv", index=False, header=['will_buy_on_return_visit'])
````
